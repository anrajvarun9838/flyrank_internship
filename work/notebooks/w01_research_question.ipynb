## 3. Quick look at the data (2-3 real numbers)

### Quick look at the starter data

The starter dataset gives an initial view of the available search-related observations. I will use these real dataset values to check whether there is enough variation in the target and available signals to justify deeper analysis over the next 7 weeks.

The numbers below are descriptive only. They do not show that any particular signal causes better search visibility.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import os
import pandas as pd
import subprocess
import sys

REPO_DIR = "/content/flyrank-ml-internship-starter"

# Clone the public starter repo if it is not already present
if not os.path.isdir(REPO_DIR):
    subprocess.run(
        [
            "git",
            "clone",
            "--depth",
            "1",
            "https://github.com/flyrank-bih/flyrank-ml-internship-starter",
            REPO_DIR,
        ],
        check=True,
    )

# Load the starter dataset
csv_path = os.path.join(
    REPO_DIR,
    "data",
    "raw",
    "content_refresh_anonymized.csv"
)

df = pd.read_csv(csv_path)

# Starter proxy label
declining = df["trend_direction"].str.lower().eq("down")

# Supporting numbers for the research question
total_pages = len(df)
declining_pages = declining.sum()
declining_rate = declining.mean()
visible_declining_pages = (
    declining & (df["impressions_90d"] >= 500)
).sum()

print(f"Total pages: {total_pages:,}")
print(f"Declining pages: {declining_pages:,}")
print(f"Declining rate: {declining_rate:.1%}")
print(f"Declining pages with >=500 impressions: {visible_declining_pages:,}")
```

##3. Quick look at the data (2-3 real numbers)
The starter dataset contains 30,000 content pages. Using the starter proxy label, 16,262 pages (54.2%) are classified as declining.

Among those declining pages, 9,961 have at least 500 impressions in the 90-day window. This suggests that there is a substantial set of visible pages where prioritizing human review could be useful.

These numbers make the refresh and content-opportunity problem worth investigating further, while the next stages will test whether better ranking methods can improve prioritization.
